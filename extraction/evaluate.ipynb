# Double Evaluation

```python
e1 = Evaluator(
    Path.cwd() / "experiments" / "FINAL/pro-low-base-persona-requirements-edge-security-focus2.reps".strip(),overwrite_training="experiments/FINAL/test.yaml"
)
e2 = Evaluator(
    Path.cwd() / "experiments" / "FINAL/pro-low-base-persona-requirements-edge-security-focus2.2".strip(),overwrite_training="experiments/FINAL/test.yaml"
)

df1 = e1.evaluate_items()
df2 = e2.evaluate_items()
df1["File"] = [f"{i}. " for i in range(len(df1))] + df1["File"].str.split("/").str[-1]
df2["File"] = [f"{i}. " for i in range(len(df2))] + df2["File"].str.split("/").str[-1]
df1['Experiment'] = e1.experiment_path.name
df2['Experiment'] = e2.experiment_path.name

# 2. Stack vertically and create the sorted MultiIndex
combined_df = pd.concat([df1, df2], axis=0)
multiview_df = combined_df.set_index(['File', 'Experiment']).sort_index()

exp1_name = e1.experiment_path.name
exp2_name = e2.experiment_path.name
exp1_temp = multiview_df.xs(exp1_name, level='Experiment')
exp2_temp = multiview_df.xs(exp2_name, level='Experiment')
exp1_temp, exp2_temp = exp1_temp.align(exp2_temp)

# # Find rows where at least one column differs
# changes_mask = (exp1_temp != exp2_temp) & ~(exp1_temp.isna() & exp2_temp.isna())
# changed_files = changes_mask.any(axis=1)

# # Overwrite multiview_df with the filtered version
# multiview_df = multiview_df.loc[changed_files.index[changed_files]]

# 3. Define the styling function for highlighting and borders
def style_multiview(df,exp1_name,exp2_name):
    # Create an empty DataFrame of the exact same shape to hold CSS strings
    styles = pd.DataFrame('', index=df.index, columns=df.columns)
    
    # Safely extract the two experiments
    exp1 = df.xs(exp1_name, level='Experiment')
    exp2 = df.xs(exp2_name, level='Experiment')
    
    # Align indices just in case a file is missing in one of the experiments
    exp1, exp2 = exp1.align(exp2)
    
    # Create a boolean mask of where the values differ
    diff_mask = (exp1 != exp2) & ~(exp1.isna() & exp2.isna())
    
    # Define our CSS rules
    border_css = 'border-bottom: 2px solid black; '
    highlight_css = 'background-color: #ffe5b4; color: #856404; '
    
    for col in df.columns:
        # A. Apply the black bottom border to ALL 'pretagging-flash-high' rows
        # This creates the visual separator between files
        exp2_rows = df.index[df.index.get_level_values('Experiment') == exp2_name]
        styles.loc[exp2_rows, col] = border_css
        
        # B. Find the specific files where this column changed
        changed_files = diff_mask.index[diff_mask[col]]
        
        # Map those files back to the MultiIndex tuple for the new experiment
        # (Filtering to ensure the file actually exists in the dataframe)
        idx_to_color = [(file, exp2_name) for file in changed_files if (file, exp2_name) in df.index]
        
        # Append the highlight CSS to the border CSS so we keep both!
        if idx_to_color:
            styles.loc[idx_to_color, col] += highlight_css
            
    return styles

# # 4. Apply the style
# multiview_df = multiview_df.sort_index(
#     level='File', 
#     key=lambda idx: idx.str.extract(r'^(\d+)', expand=False).astype(int)
# )
styled_df = multiview_df.style.apply(style_multiview, axis=None,**{"exp1_name": e1.experiment_path.name, "exp2_name": e2.experiment_path.name})
#styled_df = styled_df.hide(subset="Subspan,Superspan,Exact,Partial,IoU,Size,MultiMatch".split(","), axis="columns")
styled_df
```

# False Positives

```python
def getfps(glob):
    data = []
    for experiment in glob:
        try:
            data.extend(
                Evaluator(experiment).false_positives()
            )
        except ValueError as e:
            pass

    fps_by_doc = defaultdict(list)
    for doc_id, extraction in data:
        fps_by_doc[doc_id].append(extraction.extraction_text)

    # rapidfuzz scores from 0 to 100. 85.0 is a solid starting point for "fairly close"
    similarity_threshold = 85.0 
    clustered_results = []

    for doc_id, texts in fps_by_doc.items():
        clusters = [] 
        
        for text in texts:
            placed = False
            for cluster in clusters:
                # fuzz.ratio compares standard string similarity. 
                score = fuzz.ratio(text, cluster[0])
                
                if score >= similarity_threshold:
                    cluster.append(text)
                    placed = True
                    break
            
            if not placed:
                clusters.append([text])
                
        # 4. Compile the clusters into structured rows
        for cluster in clusters:
            clustered_results.append({
                "File": doc_id.split("/")[-1],
                "Representant": cluster[0],
                "Count": len(cluster),
                "Variants": cluster # Keeps all texts in case you need to inspect the differences
            })

    # 5. Convert to a DataFrame for analysis
    df_clusters = pd.DataFrame(clustered_results)

    # Sort by the most common false positives first
    return df_clusters.sort_values(by="Count", ascending=False).reset_index(drop=True)

def getfns(glob):
    data = []
    for experiment in glob:
        try:
            data.extend(
                Evaluator(experiment).false_negatives()
            )
        except ValueError as e:
            pass

    fps_by_doc = defaultdict(list)
    for doc_id, extraction in data:
        fps_by_doc[doc_id].append(extraction.extraction_text)

    # rapidfuzz scores from 0 to 100. 85.0 is a solid starting point for "fairly close"
    similarity_threshold = 85.0 
    clustered_results = []

    for doc_id, texts in fps_by_doc.items():
        clusters = [] 
        
        for text in texts:
            placed = False
            for cluster in clusters:
                # fuzz.ratio compares standard string similarity. 
                score = fuzz.ratio(text, cluster[0])
                
                if score >= similarity_threshold:
                    cluster.append(text)
                    placed = True
                    break
            
            if not placed:
                clusters.append([text])
                
        # 4. Compile the clusters into structured rows
        for cluster in clusters:
            clustered_results.append({
                "File": doc_id.split("/")[-1],
                "Representant": cluster[0],
                "Count": len(cluster),
                "Variants": cluster # Keeps all texts in case you need to inspect the differences
            })

    # 5. Convert to a DataFrame for analysis
    df_clusters = pd.DataFrame(clustered_results)

    # Sort by the most common false positives first
    return df_clusters.sort_values(by="Count", ascending=False).reset_index(drop=True)

```
