```python
import langextract as lx 
import ipywidgets as widgets
from IPython.display import display, HTML

def count_lines(filepath):
    with open(filepath, 'r', encoding='utf-8') as f:
        return sum(1 for _ in f)
```

```python
target = "FINAL/pro-low-base-persona-requirements-edge-security-focus2"
stage = "test"
file_path = f"/workspaces/github-best-pratices/extraction/experiments/{target}/outputs/analysis-{stage}.jsonl"

max_index = count_lines(file_path) - 1

# 2. Define the function that will run when the slider changes
def show_visualization(index):
    html_content = lx.visualize(file_path, index=index)
    display(html_content)

# 3. Create the interactive slider
# We set the initial value to 4 to match your original script
widgets.interact(
    show_visualization, 
    index=widgets.IntSlider(min=0, max=max_index, step=1, value=4, description="JSONL Index:",continuous_update=False)
);
print()
```
