```python
import pandas as pd
from collections import defaultdict
```

```python
df = pd.read_csv("OWASP & NIST - Sheet1.csv",names=["id","recommendation"])
```

```python
topics = df[df.id.str.contains("-").__invert__()]
recommendations = dict()
for row in topics.itertuples():
    key = f"{row.id}"
    recommendations[key] = dict(title=row.recommendation.strip(),content=list())

recommendations
```

```python
values = df[df.id.str.contains("-")]
for row in values.itertuples():
    key = row.id.split("-")[0].strip()
    recommendations[key]["content"].append(
        (row.id,row.recommendation.strip())
    )

recommendations
```

```python
for key,val in recommendations.items():
    df = pd.DataFrame(data=val["content"],columns=["ID","Recommedation"])
    df.index = df.ID
    val["content"] = df.to_markdown(index=False)
```

```python
with open("OWASPT10.md","w") as f:
    f.writelines("\n".join([
        "# OWASP Top 10 CI/CD Security Risks",
        "",
        "Each security risk have several recommendations which have an ID, e.g. 'SEC01-01', and a recommendation afterwards",
        "",
    ]))
    for reco,content in recommendations.items():
        f.writelines("\n".join([
            "",
            "",
            f"## {reco} - {content['title']}",
            "",
            f"{content['content']}"
            "",
        ]))
```
